```python
import numpy as np
import pandas as pd 
```

```python
df = pd.read_csv('fra_cleaned_with_images.csv', encoding='latin-1', on_bad_lines='skip', delimiter=';')
df.head()
```

```python
df.rename(columns={'ï»¿url': 'url'}, inplace=True)
df.head()
```

```python
print(df.columns)
```

```python
columns_to_replace = ["perfume", "brand", "top", "middle", "base"]
df[columns_to_replace] = df[columns_to_replace].replace("-", " ", regex=True)
```

```python
df.head()
```

```python
def to_set(value):
    if value is None:
        return set()

    text = str(value).strip().lower()
    if not text or text == "nan":
        return set()

    return {item.strip() for item in text.split(",") if item.strip()}

ACCORD_COLUMNS = ["mainaccord1", "mainaccord2", "mainaccord3", "mainaccord4", "mainaccord5"]

def accords_from_row(row):
    accords = set()
    for col in ACCORD_COLUMNS:
        value = row[col]
        if value is not None:
            text = str(value).strip().lower()
            if text and text != "nan":
                accords.add(text)
    return accords

def normalize_gender(value):
    text = str(value).strip().lower()

    if text in {"male", "man", "men", "masculine"}:
        return "male"
    if text in {"female", "woman", "women", "feminine"}:
        return "female"
    if text in {"unisex", "shared"}:
        return "unisex"

    return "unknown"

perfumes = []

for idx, row in df.iterrows():
    perfumes.append({
        "id": idx,
        "perfume": row["perfume"],
        "brand": row["brand"],
        "top": to_set(row["top"]),
        "middle": to_set(row["middle"]),
        "base": to_set(row["base"]),
        "accords": accords_from_row(row),
        "gender": normalize_gender(row["gender"]),
        "rating_value": float(row["rating value"]) if str(row["rating value"]) != "nan" else 0.0,
        "rating_count": int(row["rating count"]) if str(row["rating count"]) != "nan" else 0,
        "url": row["url"],
        "image_url": row["image_url"]
    })
```

```python
perfumes[1]
```

```python
#Build vocabularies from perfume data

all_notes = set()
all_accords = set()

for p in perfumes:
    all_notes.update(p["top"])
    all_notes.update(p["middle"])
    all_notes.update(p["base"])
    all_accords.update(p["accords"])

all_notes = sorted(all_notes)
all_accords = sorted(all_accords)
```

```python
import json

# Convert sets to lists for JSON serialization
perfumes_json = []
for p in perfumes:
    p_copy = p.copy()
    p_copy['top'] = list(p_copy['top'])
    p_copy['middle'] = list(p_copy['middle'])
    p_copy['base'] = list(p_copy['base'])
    p_copy['accords'] = list(p_copy['accords'])
    perfumes_json.append(p_copy)

with open('perfumes.json', 'w') as f:
    json.dump(perfumes_json, f, indent=2)
```
